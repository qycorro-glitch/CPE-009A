```python
# 1. Make a program that would calculate the distance between two points given a list of coordinates. Use the distance formula.
#coorindates_list = [(1,1), (2,3)]

coordinates_list = [(1, 1), (2, 3)]

x1 = coordinates_list[0][0]
y1 = coordinates_list[0][1]

x2 = coordinates_list[1][0]
y2 = coordinates_list[1][1]

distance = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
print("The distance between the two points is:", distance)

```

```python
#2. For a given string input, replace all the words “stupid” with an asterisk * equal to the length of the string. The new
#string value should be displayed with the asterisks.
# Phonebook

text = input("Enter a sentence: ")
bad_word = "stupid"
stars = "*" * len(bad_word)

result = text.replace(bad_word, stars)
result = result.replace(bad_word.capitalize(), stars)
result = result.replace(bad_word.upper(), stars)

print("Updated sentence:")
print(result)
```

```python
#3.3. Create a simple phonebook program that can read from a
#list of dictionaries. The program should be able to display a
#person’s name, contact, and address based from a user input which is
#the id of the record.
phonebook = [
    {"id": 1, "name": "Alice", "contact": "123-456-7890", "address": "123 Main St"},
    {"id": 2, "name": "Bob", "contact": "987-654-3210", "address": "456 Elm St"},
    {"id": 3, "name": "Charlie", "contact": "555-555-5555", "address": "789 Oak St"}
]


user_id = int(input("Enter the ID of the contact you want to find: "))


found = False


for person in phonebook:
    if person["id"] == user_id:
        print("Name:", person["name"])
        print("Contact:", person["contact"])
        print("Address:", person["address"])
        found = True
        break


if not found:
    print("No contact found with that ID.")
```
