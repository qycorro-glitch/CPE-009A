```python
class Car:
    def __init__(self, name, mileage):
        self.name = name
        self.mileage = mileage

    def description(self):
        return f"The {self.name} car gives the mileage of {self.mileage}km/l"

obj1 = Car("Toyota", 2202)
obj2 = Car("Subaru", 1200)

print(obj1.description())
print(obj2.description())
```

```python
class Animals:
  def __init__(self,name,habitat,temperature):
    self.name = name
    self.habitat = habitat
    self.temperature = temperature


  def features(self):
    return f"The {self.name} is a {self.temperature}-blooded animal that lives on {self.habitat}."

mammals = Animals("Dog", "Land", "Warm")
birds = Animals("Sparrow", "Air", "Warm")
reptiles = Animals("Snake", "Terrestrial", "Cold")

print(mammals.name)
print(mammals.habitat)

print(birds.name)
print(birds.habitat)

print(reptiles.name)
print(reptiles.habitat)



print(mammals.features())
print(birds.features())
print(reptiles.features())
```

```python
class Shape:
    def __init__(self, name, sides):
        self.name = name
        self.sides = sides

    def area(self):
        return "Area formula not defined for generic shape."

class Triangle(Shape):
    def __init__(self, base, height):
        super().__init__("Triangle", 3)
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle", 0)
        self.radius = radius

    def area(self):
        return 3.1416 * (self.radius ** 2)

class Square(Shape):
    def __init__(self, side):
        super().__init__("Square", 4)
        self.side = side

    def area(self):
        return self.side ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__("Rectangle", 4)
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


triangle = Triangle(10, 5)
circle = Circle(7)
square = Square(4)
rectangle = Rectangle(6, 3)

print(triangle.name, "Area:", triangle.area())
print(circle.name, "Area:", circle.area())
print(square.name, "Area:", square.area())
print(rectangle.name, "Area:", rectangle.area())
```
